# One-loop galaxy power spectrum: Lagrangian PT (`LPT`)

```python
import jax
jax.config.update('jax_enable_x64', True)
import jax.numpy as jnp
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.family'] ='sans-serif'
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams['xtick.major.size'] = 6.0
plt.rcParams['ytick.major.size'] = 6.0
plt.rcParams['xtick.major.width'] = 1.0
plt.rcParams['ytick.major.width'] = 1.0
plt.rcParams['xtick.minor.size'] = 4.0
plt.rcParams['ytick.minor.size'] = 4.0
plt.rcParams['xtick.minor.width'] = 0.8
plt.rcParams['ytick.minor.width'] = 0.8
plt.rcParams['xtick.top'] = True
plt.rcParams['ytick.right'] = True
plt.rcParams['xtick.minor.visible'] = True
plt.rcParams['ytick.minor.visible'] = True
plt.rcParams['font.size'] = 16
plt.rcParams['axes.linewidth'] = 1.5
plt.rcParams['text.usetex'] = False
```

```python
import elf
import elf.params as ps_params
```

## Model setup

`counterterm_base` selects the spectrum the counterterms ride on:

| value | counterterm base |
|---|---|
| `'linear_ir_resum'` (default) | the IR-resummed linear power spectrum |
| `'linear'` | the linear power spectrum without the IR resummation |
| `'zeldovich'` | the Zel'dovich power spectrum |

```python
lpt = elf.LPT(counterterm_base='linear_ir_resum')
```

## Parameters

The linear matter power spectrum is passed as an array of shape `(2, nk)`
stacking $k$ and $P_{\rm lin}(k)$.  The EFT parameters are grouped into

| group | entries |
|---|---|
| `bias`  | $b_{1}^{\rm L},\ b_{2}^{\rm L},\ b_{\mathcal{G}_2}^{\rm L},\ b_{\Gamma_3}^{\rm L}$ (Lagrangian bias)|
| `ctr`   | $c_0,\ c_2,\ c_4,\ c_6,\ c_{44},\ c_{46},\ c_{48}$ |
| `stoch` | $P_{\rm shot},\ a_0,\ a_2$ |

```python
## input linear matter power spectrum
d = np.loadtxt('data/pk_lin_planck18_fid_z0.txt')
pk_data = jnp.stack([d[:,0], d[:,1]], axis=0)

## other model parameters
f = 0.578
h = 0.6736
bias  = jnp.array([1.0, -0.5, 0.09, 0.15])        # b1L, b2L, bG2L, bGamma3L
ctr   = jnp.array([5.0, 10.0, -5.0, 0.0, 100.0, 0.0, 0.0])  # c0, c2, c4, c6, c44, c46, c48
stoch = jnp.array([1000.0, 0.0, 0.0])             # e00, e20, e22

params = ps_params.make_params(f=f, h=h, bias=bias, ctr=ctr, stoch=stoch)
```

## Redshift-space $P(k, \mu)$

```python
## Output P(k,mu) as jax.numpy array of shape (len(k), len(mu))
k = jnp.linspace(1e-3, 0.3, 100)
mu = jnp.linspace(0., 1., 6)

lpt.get_pkmu(k, mu, pk_data, params).block_until_ready()   # trigger JIT compilation
%time pkmu = lpt.get_pkmu(k, mu, pk_data, params).block_until_ready()
```

```python
fig = plt.figure(figsize=(7,5))
color = plt.rcParams['axes.prop_cycle'].by_key()['color']
ax = fig.add_subplot()

for i, mu_i in enumerate(mu):
    ax.plot(k, k * pkmu[:,i], c=color[i], label=r'$\mu = {%.1f}$' % (mu_i))

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P(k, \mu) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=12, bbox_to_anchor=(1.01, 1.0), loc='upper left')
plt.grid(linestyle='--')
```

## Redshift-space multipoles ($\ell = 0, 2, 4$ are supported)

```python
lpt.get_pk_ells(k, pk_data, params).block_until_ready()   # trigger JIT compilation
%time pkl = lpt.get_pk_ells(k, pk_data, params).block_until_ready()
```

```python
fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot()

for i, l in enumerate([0,2,4]):
    ax.plot(k, k * pkl[i], label=r'$\ell = %s$' % (l))

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P_{\ell}(k) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=16)
plt.grid(linestyle='--')
```

### Changing the parameters

Call the same object again with new `params` (or a new `pk_data`); nothing is recompiled.

```python
params2 = elf.make_params(f=0.55, h=0.70, bias=jnp.array([0.8, -0.3, 0.05, 0.1]),
                           ctr=jnp.array([0.0] * 7), stoch=jnp.array([0.0, 0.0, 0.0]))
pkl2 = lpt.get_pk_ells(k, pk_data, params2)

fig = plt.figure(figsize=(7,5))
color = plt.rcParams['axes.prop_cycle'].by_key()['color']
ax = fig.add_subplot()

for i, l in enumerate([0,2,4]):
    ax.plot(k, k * pkl[i], c=color[i], label=r'$\ell = %s$' % (l))
    ax.plot(k, k * pkl2[i], c=color[i], ls='--')

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P_{\ell}(k) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=16, bbox_to_anchor=(1.01, 1.0), loc='upper left')
plt.grid(linestyle='--')
```

### Cross power spectrum

Give the second tracer's (Lagrangian) bias as `bias2`; `ctr` and `stoch` are the coefficients of the cross spectrum itself. Counterterms built from each tracer's own coefficients, `[Z1^A c^B(mu) + Z1^B c^A(mu)] k^2 P`, are a polynomial in `mu^2` and are passed by matching its coefficients; the cross shot noise is a parameter of its own.

```python
params_ab = elf.make_params(
    f=f, h=h,
    bias=bias,                                       # tracer a
    bias2=jnp.array([0.35, 0.05, 0.2, -0.05]),       # tracer b
    ctr=jnp.array([0.7, 1.5, -0.7, 0.0, 7.0, 0.0, 0.0]),
    stoch=jnp.array([0.0, 0.0, 0.0]),                # cross shot noise (0 for disjoint samples)
)
pkl_ab = lpt.get_pk_ells(k, pk_data, params_ab)

fig = plt.figure(figsize=(7,5))
color = plt.rcParams['axes.prop_cycle'].by_key()['color']
ax = fig.add_subplot()

for i, l in enumerate([0,2,4]):
    ax.plot(k, k * pkl[i], c=color[i], label=r'$\ell = %s$' % (l))
    ax.plot(k, k * pkl_ab[i], c=color[i], ls='--')

ax.text(0.96, 0.9, 'solid: auto, dashed: cross', fontsize=13, transform=ax.transAxes, ha='right')

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P_{\ell}(k) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=16, bbox_to_anchor=(1.01, 1.0), loc='upper left')
plt.grid(linestyle='--')
```

## Alcock-Paczynski effect

```python
# AP parameters, depending on the reference cosmology and redshift
alpha_perp = 1.01
alpha_para = 1.0

lpt.get_pk_ells(k, pk_data, params, alpha_perp=alpha_perp, alpha_para=alpha_para).block_until_ready()
%time pkl_AP = lpt.get_pk_ells(k, pk_data, params, alpha_perp=alpha_perp, alpha_para=alpha_para).block_until_ready()
```

```python
fig = plt.figure(figsize=(7,5))
color = plt.rcParams['axes.prop_cycle'].by_key()['color']
ax = fig.add_subplot()

for i, l in enumerate([0,2,4]):
    ax.plot(k, k * pkl_AP[i], c=color[i], label=r'$\ell = %s$' % (l))
    ax.plot(k, k * pkl[i], ls='--', c=color[i])

ax.text(0.96, 0.9, r'$\alpha_\perp=%s, \alpha_\parallel=%s$' % (alpha_perp, alpha_para),
        fontsize=16, transform=ax.transAxes, ha='right')

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P_{\ell}(k) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=16, bbox_to_anchor=(1.01, 1.0), loc='upper left')
plt.grid(linestyle='--')
```
