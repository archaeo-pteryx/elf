## Model setup

`counterterm_base` selects the spectrum the counterterms ride on:

| value | counterterm base |
|---|---|
| `'zeldovich'` (default) | the Zel'dovich integrand, fused into the same Hankel transform |
| `'linear'` | the linear $P(k)$, added in $k$ space |
| `'linear_ir_resum'` | the IR-resummed linear $P(k,\mu)$ (needs `h` in the parameters) |

```python
lpt = ps_1loop_jax.LPT(counterterm_base='linear_ir_resum')
```

## Parameters

The linear matter power spectrum is passed as an array of shape `(2, nk)`
stacking $k$ and $P_{\rm lin}(k)$.  The EFT parameters are grouped into

| group | entries |
|---|---|
| `bias`  | $b_{1}^{\rm L},\ b_{2}^{\rm L},\ b_{\mathcal{G}_2}^{\rm L},\ b_{\Gamma_3}^{\rm L}$ (Lagrangian bias)|
| `ctr`   | $c_0,\ c_2,\ c_4,\ c_6,\ c_{\rm nlo}$ |
| `stoch` | $P_{\rm shot},\ a_0,\ a_2$ |

`h` is only needed for `counterterm_base='linear_ir_resum'`.

```python
## input linear matter power spectrum
d = np.loadtxt('data/pk_lin_planck18_fid_z0.txt')
pk_data = jnp.stack([d[:,0], d[:,1]], axis=0)

## other model parameters
f = 0.578
bias  = jnp.array([0.71, 0.26, 0.67, -0.2])     # b1_L, b2, bG2, bGamma3
ctr   = jnp.array([1.0, 2.0, -1.0, 0.0, 10.0])  # c0, c2, c4, c6, c_nlo
stoch = jnp.array([3333.0, 0.0, 0.0])           # P_shot, a0, a2

params = ps_params.make_lpt_params(f=f, bias=bias, ctr=ctr, stoch=stoch, h=0.6736)
```

## Redshift-space $P(k, \mu)$

```python
## Output P(k,mu) as jax.numpy array of shape (len(k), len(mu))
k = jnp.linspace(1e-3, 0.3, 100)
mu = jnp.linspace(0., 1., 6)

lpt.get_pkmu(k, mu, pk_data, params).block_until_ready()   # trigger JIT compilation
%time pkmu = lpt.get_pkmu(k, mu, pk_data, params).block_until_ready()
```

```python
fig = plt.figure(figsize=(7,5))
color = plt.rcParams['axes.prop_cycle'].by_key()['color']
ax = fig.add_subplot()

for i, mu_i in enumerate(mu):
    ax.plot(k, k * pkmu[:,i], c=color[i], label=r'$\mu = {%.1f}$' % (mu_i))

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P(k, \mu) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=12)
plt.grid(linestyle='--')
```

## Cross power spectrum of two tracers

As in the EPT backend, passing a second parameter set turns the auto spectrum
into a cross spectrum.  Bias monomials are symmetrised ($b_1^2 \to b_1^a b_1^b$,
$b_1 b_2 \to (b_1^a b_2^b + b_1^b b_2^a)/2$, ...) and the counterterm
coefficients are averaged.  Cross shot noise is passed explicitly via `stoch=`.

```python
params_a = params
params_b = ps_params.make_lpt_params(
    f=f,
    bias=jnp.array([0.35, 0.05, 0.2, -0.05]),
    ctr=jnp.array([0.5, 1.0, -0.5, 0.0, 5.0]),
    stoch=jnp.array([5000.0, 0.0, 0.0]),
    h=0.6736,
)

stoch_ab = jnp.array([0.0, 0.0, 0.0])   # disjoint samples -> no cross shot noise

pkl_aa = model.get_pk_ells(k, pk_data, params_a)
pkl_bb = model.get_pk_ells(k, pk_data, params_b)
model.get_pk_ells(k, pk_data, params_a, params_b, stoch=stoch_ab).block_until_ready()
%time pkl_ab = model.get_pk_ells(k, pk_data, params_a, params_b, stoch=stoch_ab).block_until_ready()
```

```python
fig = plt.figure(figsize=(7,5))
color = plt.rcParams['axes.prop_cycle'].by_key()['color']
ax = fig.add_subplot()

for i, (lab, p) in enumerate([('$P_{aa}$', pkl_aa), ('$P_{bb}$', pkl_bb), ('$P_{ab}$', pkl_ab)]):
    ax.plot(k, k * p[0], c=color[i], label=lab + r'$,\ \ell=0$')
    ax.plot(k, k * p[1], c=color[i], ls='--', label=lab + r'$,\ \ell=2$')

plt.xlabel(r'$k \, [h \, \mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$k \, P_{\ell}(k) \, [h^{-2} \, \mathrm{Mpc}^2]$')
plt.xlim(k.min(), k.max())
plt.legend(fontsize=12, ncol=3, bbox_to_anchor=(1.01, 1.0), loc='upper left')
plt.grid(linestyle='--')
```
